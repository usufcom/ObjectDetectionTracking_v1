## Reading, writing and displaying images with OpenCV

Let's start by importing the OpenCV libary 

```python
# Press Shift + ENTER to run this line
# You should see an * between the [ ] on the left
# OpenCV takes a couple seconds to import the first time

import cv2 # OpenCV-Python
import numpy as np
#%matplotlib inline
import matplotlib.pyplot as plt

#print("Python Version %s" % python.__version__)
print("OpenCV-Python Version %s" % cv2.__version__)
```

Let's now load our first image

```python
# We don't need to do this again, but it's a good habit
#import cv2 

# Load an image using 'imread' specifying the path to image
image = cv2.imread(r'images/Einstein.jpg')

# Convert from BGR to RGB
im_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Our file 'input.jpg' is now loaded and stored in python 
# as a varaible we named 'image'

# Show image
plt.imshow(im_rgb)
plt.title('RGB Image')
plt.show()
```

### Let's take a closer look at how images are stored

```python
print(im_rgb.shape)
```

#### Shape gives the dimensions of the image array

The 2D dimensions are 830 pixels in high bv 1245 pixels wide.
The '3L' means that there are 3 other components (RGB) that make up this image.

```python
# Let's print each dimension of the image

print ('Height of Image:', int(im_rgb.shape[0]), 'pixels')
print ('Width of Image: ', int(im_rgb.shape[1]), 'pixels')
print ('Depth(chanels) of Image: ', int(im_rgb.shape[2]), 'chanels')
```

## Reading and displaying videos with OpenCV

```python
import numpy as np
import cv2 # OpenCV-Python
#%matplotlib inline
import matplotlib.pyplot as plt
import time
```

```python
# Open a video with cv2.VideoCapture
vid = cv2.VideoCapture(r"images/walking.avi")

#get the fps of the video
fps = int(vid.get(cv2.CAP_PROP_FPS))
print('The Frame Rate is:',fps)
Vid_Speed = 1
fps1 = int(fps*Vid_Speed)
print('fsp1:',fps1)
```

#### Read video file and display the video

```python
# Open a video with cv2.VideoCapture
vid = cv2.VideoCapture(r"images/walking.avi")

#get the fps of the video
fps = int(vid.get(cv2.CAP_PROP_FPS))

while True:
    # Capture frame-by-frame
    ret, frame = vid.read()

    if not ret:
        print("Can't receive frame (stream end). Exiting ...")
        break    
    # Display the resulting frame
    cv2.imshow('frame', frame)
    # delay for fps
    time.sleep(1/fps)
    if cv2.waitKey(1) == ord('q'):
        break
# When everything done, release the capture
vid.release()
cv2.destroyAllWindows()
```
